```python
import pandas as pd
import numpy as np

# قراءة البيانات للبدء بهندسة الخصائص
df = pd.read_parquet('orders_merged.parquet')

print("حجم البيانات الأساسية:", df.shape)

# تحويل أعمدة التواريخ إلى صيغة datetime لتسهيل استخراج الخصائص الزمنية
date_columns = ['order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date', 'order_delivered_customer_date']

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col])

print("تم تحويل التواريخ بنجاح. جاهزون لاستخراج الخصائص الزمنية.")
df.head(2)
```
