```python
import pandas as pd
from sklearn.model_selection import train_test_split

# قراءة البيانات المعالجة التي تم تجهيزها سابقاً
df = pd.read_parquet('orders_merged.parquet')

print("حجم البيانات الكلي قبل التقسيم:", df.shape)

# تقسيم البيانات بنجاح (مثلاً 80% تدريب و 20% اختبار أو حسب متطلبات مشروعك)
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

print("عدد صفوف بيانات التدريب (Train):", train_df.shape)
print("عدد صفوف بيانات الاختبار (Test):", test_df.shape)
```
